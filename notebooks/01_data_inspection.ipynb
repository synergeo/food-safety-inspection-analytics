```python
# Calculate missing values across the complete dataset using chunks

missing_counts = None
total_rows_checked = 0

for chunk in pd.read_csv(
    file_path,
    sep=";",
    encoding="latin1",
    chunksize=100_000,
    low_memory=False
):
    # Count missing values in the current chunk
    chunk_missing = chunk.isnull().sum()

    if missing_counts is None:
        missing_counts = chunk_missing
    else:
        missing_counts = missing_counts.add(chunk_missing, fill_value=0)

    total_rows_checked += len(chunk)

# Create the full-dataset missing-value summary
missing_summary_full = pd.DataFrame({
    "Missing_Count": missing_counts.astype(int),
    "Missing_Percentage": (
        missing_counts / total_rows_checked * 100
    ).round(2)
})

# Keep only columns with missing values
missing_summary_full = missing_summary_full[
    missing_summary_full["Missing_Count"] > 0
].sort_values(
    by="Missing_Percentage",
    ascending=False
)

print(f"Total rows analysed: {total_rows_checked:,}")
missing_summary_full
```

```python
# Count MatrixGrT (Matrix Group Description) across the complete dataset

matrix_group_counts = {}

for chunk in pd.read_csv(
    file_path,
    sep=";",
    encoding="latin1",
    usecols=["MatrixGrT"],
    chunksize=100_000,
    low_memory=False
):
    counts = chunk["MatrixGrT"].value_counts(dropna=False)

    for category, count in counts.items():
        matrix_group_counts[category] = matrix_group_counts.get(category, 0) + count

matrix_group_summary = (
    pd.Series(matrix_group_counts)
    .sort_values(ascending=False)
    .to_frame(name="Count")
)

matrix_group_summary["Percentage"] = (
    matrix_group_summary["Count"] / total_rows_checked * 100
).round(2)

matrix_group_summary
```

---

## Implication for Target Definition

The distribution of "BewertungT" (Assessment Description) provides useful information for the Failure-Risk Prediction objective, but a binary target should **not yet be created solely from these labels**.

Three issues require further investigation.

First, the assessment distribution is extremely imbalanced. Explicit adverse assessment categories represent only a very small fraction of the more than 2.15 million analytical records.

Second, some categories describe analytical or regulatory conditions rather than a simple compliant/non-compliant outcome. For example, "Stoffnachweis ist bei der entsprechenden Summe zu bewerten." (Detection of the substance must be assessed using the corresponding sum) does not by itself provide a straightforward failure classification.

Third, some apparently similar categories have different regulatory meanings. For example, "> Höchstmenge (ohne Beanstandung)" (Above maximum permitted level, without non-compliance) must be distinguished from "> Höchstmenge (führt zur Beanstandung)" (Above maximum permitted level, leading to non-compliance).

Therefore, the meaning and structure of "Bewertung" (Assessment Code) and "BewertungT" (Assessment Description) must be investigated further before constructing the final target variable.

---

## Current Interpretation

This analysis provides evidence that "BewertungT" (Assessment Description) contains potentially valuable outcome information for the project.

However, the results also demonstrate that **Failure-Risk Prediction cannot simply be defined as "above a limit = failure."**

The eventual target definition must distinguish between:

- compliant or acceptable assessments;
- explicit non-compliance assessments;
- assessments requiring additional interpretation; and
- records for which no assessment value or determination is available.

This distinction will help ensure that the target variable represents a meaningful food-safety outcome and reduces the risk of incorrectly labelling observations during the modelling stage.

# Chi-Square Assumption Check for Country of Origin

A chi-square test of independence was initially considered to evaluate whether "HerkStaatT" (Country of Origin Description) is associated with the preliminary adverse target.

The hypotheses were:

**Null hypothesis (H₀):**  
Country of origin and adverse outcome are independent.

**Alternative hypothesis (H₁):**  
Country of origin and adverse outcome are associated.

## Initial Test Result

- Chi-square statistic: **103.74**
- p-value: **0.093483**
- Degrees of freedom: **86**

However, the assumptions of the chi-square approximation were also examined.

Of the **174 expected contingency-table cells**, **114 cells (65.52%)** had an expected frequency below 5.

## Interpretation

The contingency table is highly sparse because there are **87 origin categories** but only **71 adverse food samples**.

A large proportion of expected cell frequencies therefore falls below the level normally considered suitable for a standard chi-square approximation.

Consequently, the p-value of **0.093483 should not be treated as a reliable inferential conclusion** for the original 87-category country variable.

The test result is retained for transparency, but no statistical conclusion about independence or association will be drawn from this version of the chi-square test.

## Next Statistical Approach

A more defensible analysis requires reducing sparse origin categories into meaningful broader groups before repeating the association test.

This grouping should be determined from interpretable characteristics rather than from adverse outcomes themselves, in order to avoid target leakage or outcome-driven category construction.

# Resolution of Multiple Country-of-Origin Values

The sample-level consistency assessment identified only **2 of 8,582 food-scope samples** with multiple values for "HerkStaatT" (Country of Origin Description).

The two cases were:

- Sample 3692: "Vietnam" (Vietnam) and "Ungeklärt" (Unresolved/Unknown)
- Sample 7835: "Indien, einschl. Sikkim und Goa" (India, including Sikkim and Goa) and "Ungeklärt" (Unresolved/Unknown)

## Data-Cleaning Decision

In both cases, a specific country of origin is available alongside "Ungeklärt" (Unresolved/Unknown).

The specific country is therefore retained as the sample-level country-of-origin value.

This approach prioritizes the available informative value without introducing information that is not present in the original dataset.

No food samples are removed as a result of this issue.

```python
# Create one Country of Origin feature per sample

def resolve_country(values):
    values = list(values.dropna().unique())

    # If a specific country exists together with
    # Ungeklärt (Unresolved/Unknown), keep the specific country
    specific_values = [
        value for value in values
        if value != "Ungeklärt"
    ]

    if len(specific_values) == 1:
        return specific_values[0]

    if len(values) == 1:
        return values[0]

    if len(values) == 0:
        return "Missing"

    # Safety fallback if another unexpected combination exists
    return " | ".join(sorted(values))


country_feature = (
    sample_features_raw[
        ["ProbenNrPseud", "HerkStaatT"]
    ]
    .drop_duplicates()
    .groupby("ProbenNrPseud")["HerkStaatT"]
    .apply(resolve_country)
    .reindex(sorted(food_sample_ids))
    .reset_index(name="Country_Feature")
)

print("Number of samples:", len(country_feature))

print("\nMissing after processing:")
print(country_feature["Country_Feature"].isna().sum())

print("\nNumber of country categories:")
print(country_feature["Country_Feature"].nunique())

print("\nResolved multi-valued samples:")
print(
    country_feature[
        country_feature["ProbenNrPseud"].isin([3692, 7835])
    ]
)
```

## Modelling Interpretation

Each row now represents exactly one food sample rather than one analytical measurement.

This is important because the original analytical dataset contained many repeated laboratory measurements for the same sample.

Using a sample-level structure prevents samples with many analytical measurements from being disproportionately represented in the predictive model.

"ProbenNrPseud" (Pseudonymised Sample Number) is retained as an identifier for traceability but will **not be used as a predictive feature**.

Variables used directly to construct "Target_Adverse" (Adverse Target), including "Bewertung" (Assessment) and "BewertungT" (Assessment Description), are excluded from the predictor dataset to prevent target leakage.

## Class-Imbalance Consideration

The target remains severely imbalanced, with only **71 adverse samples among 8,582 food samples**.

Consequently, model evaluation will not rely on accuracy alone.

Metrics such as precision, recall, F1-score, confusion matrix results and precision-recall performance will be considered during modelling.

Class-weighting and appropriate resampling approaches will also be evaluated using the training data only.
